### Foundation Data Processing Functions

````python
import pandas as pd

def load_and_clean_transaction_data(file_path):
    # Step 1: Read the CSV file into a DataFrame
    df = pd.read_csv(file_path)

    # Step 2: Check if all required columns are present
    required_columns = ["Date", "Amount", "Category", "Description"]
    for col in required_columns:
        if col not in df.columns:
            print("❌ Error: The CSV file is missing the required column: " + col)
            return None # Stop the function early if a column is missing

    # Step 3: Handle missing values by filling them with smart defaults
    df["Category"] = df["Category"].fillna("Uncategorized")
    df["Description"] = df["Description"].fillna("No Description")
    df["Date"] = df["Date"].fillna("Unknown Date")
    df["Amount"] = df["Amount"].fillna("$0.00")

    # Step 4: Clean the Amount column (Remove '$' and ',' signs)
    df["Amount"] = df["Amount"].astype(str)
    df["Amount"] = df["Amount"].str.replace("$", "", regex=False)
    df["Amount"] = df["Amount"].str.replace(",", "", regex=False)

    # Step 5: Convert the Amount column from text to numbers
    df["Amount"] = pd.to_numeric(df["Amount"])

    print("✅ Success: Data loaded and cleaned successfully!")
    return df
````

````python
# Save sample data into CSV
df_sample.to_csv("student_transactions.csv", index=False)

# Run cleaning function
cleaned_df = load_and_clean_transaction_data("student_transactions.csv")

# Show cleaned data
print(cleaned_df.head())
````

````python
import pandas as pd

def analyze_spending_patterns(df):
    # Step 1: Separate expenses from income
    # (We only want to analyze money going OUT, which are negative numbers)
    expense_df = df[df["Amount"] < 0].copy()

    # Convert negative numbers to positive so it's easier to read on a chart/report
    expense_df["Amount"] = expense_df["Amount"].abs()

    # Step 2: Calculate total spending by category
    # .groupby() clusters identical categories, and .sum() adds up their amounts
    category_totals = expense_df.groupby("Category")["Amount"].sum()

    # Step 3: Calculate the overall total spending
    total_overall_spending = category_totals.sum()

    # Step 4: Calculate percentage spending by category
    category_percentages = (category_totals / total_overall_spending) * 100

    # Step 5: Identify the top spending category
    # .idxmax() finds the name of the category with the highest total amount
    top_category = category_totals.idxmax()
    top_category_amount = category_totals.max()

    # Step 6: Create simple business-style financial insights
    insight_1 = f"Your highest expense category is '{top_category}', totaling ${top_category_amount:.2f}."
    insight_2 = f"This single category makes up {category_percentages[top_category]:.1f}% of your total spending."

    # Step 7: Format the totals into a clean dictionary with nice dollar signs
    formatted_totals = {}
    for category, amount in category_totals.items():
        formatted_totals[category] = f"${amount:.2f}"

    # Step 8: Package everything into a final results dictionary
    analysis_results = {
        "Total Overall Spending": f"${total_overall_spending:.2f}",
        "Spending By Category": formatted_totals,
        "Top Category": top_category,
        "Financial Insights": [insight_1, insight_2]
    }

    return analysis_results
````

````python
# 1. Clean the data using your previous function
cleaned_df = load_and_clean_transaction_data("student_transactions.csv")

# 2. Run the analysis function
results = analyze_spending_patterns(cleaned_df)

# 3. Print out the results nicely
print("\n--- FINANCIAL ANALYSIS REPORT ---")
print("Total Money Spent:", results["Total Overall Spending"])

print("\nBreakdown by Category:")
for cat, amt in results["Spending By Category"].items():
    print(f" - {cat}: {amt}")

print("\nSmart Insights:")
for insight in results["Financial Insights"]:
    print(f" 💡 {insight}")
````

````python
def generate_financial_recommendations(analysis_data):
    """
    Generates a simple, actionable financial recommendation report
    based on the results of the spending analysis.
    """
    top_category = analysis_data["Top Category"]
    total_spent = analysis_data["Total Overall Spending"]

    report = "=========================================\n"
    report += "       SMART FINANCE RECOMMENDATIONS     \n"
    report += "=========================================\n\n"
    report += f"Based on your total spending of {total_spent}, here is your advice:\n\n"

    if top_category == "Groceries":
        advice = "💡 Recommendation 1: Try meal prepping for the week.\n"
        advice += "   Buying planned ingredients can help reduce unnecessary grocery spending.\n\n"
        advice += "💡 Recommendation 2: Check your fridge before shopping.\n"
        advice += "   This can reduce duplicate purchases and food waste."

    elif top_category == "Dining" or top_category == "Entertainment":
        advice = "💡 Recommendation 1: Set a weekly social spending limit.\n"
        advice += "   This helps control casual dining and entertainment costs.\n\n"
        advice += "💡 Recommendation 2: Look for student discounts or free activities.\n"
        advice += "   This keeps social spending lower without removing it completely."

    elif top_category == "Transport":
        advice = "💡 Recommendation 1: Review your transport habits.\n"
        advice += "   Check whether student concession fares or planned trips can reduce costs.\n\n"
        advice += "💡 Recommendation 2: Combine trips where possible.\n"
        advice += "   Fewer small transport transactions can reduce weekly spending."

    else:
        advice = "💡 Recommendation 1: Review your recent transactions.\n"
        advice += f"   Your biggest expense was '{top_category}', so check whether it was essential.\n\n"
        advice += "💡 Recommendation 2: Try the 50/30/20 rule.\n"
        advice += "   Put 50% toward needs, 30% toward wants, and 20% toward savings."

    report += f"🔴 Current Focus Area: Your top spending category is '{top_category}'.\n\n"
    report += "Actionable Steps to Save Money:\n"
    report += "-----------------------------------------\n"
    report += advice + "\n"
    report += "========================================="

    return report
````

````python
from hands_on_ai import rag
from hands_on_ai.chat import get_response

def run_financial_rag_assistant(user_question):
    print(f"User Asked: '{user_question}'")
    print("--------------------------------------------------")

    # Step 1: Load the raw text file
    raw_text = rag.load_text_file("student_budget_guide.txt")

    # Step 2: Chop the text into smaller, readable chunks
    # This prevents overloading the AI with an entire textbook at once
    chunks = rag.chunk_text(raw_text)

    # Step 3: Turn those chunks into mathematical vectors (embeddings)
    # This allows the system to understand the 'meaning' of the words
    embeddings = rag.get_embeddings(chunks)

    # Step 4: Find the top 1 best-matching chunk for our question
    # get_top_k searches the mathematical space and pulls out the most relevant text
    top_matches = rag.get_top_k(user_question, chunks, embeddings, k=1)

    # Extract the actual text string from our top match
    retrieved_context = top_matches[0]
    print(f"🔍 Found Relevant Source Text:\n{retrieved_context}\n")

    # Step 5: Augment the prompt by pinning the source text to the question
    augmented_prompt = f"""
    You are SaverBuddy, a helpful university financial coach.
    Answer the user's question using ONLY the facts provided in the Context below.
    If the answer isn't in the context, kindly say you don't know.

    Context:
    {retrieved_context}

    Question:
    {user_question}
    """

    # Step 6: Generate the final answer using your main chat tool
    # (Optional: Pass your system persona string if you have it active)
    ai_answer = get_response(augmented_prompt)

    return ai_answer
````

## Custom Financial Tools

````python
def create_savings_calculator_tool(current_savings, monthly_contribution, target_amount):
    """
    Calculates the number of months required to reach a specific financial savings goal.

    Inputs:
    current_savings (float): The starting balance the user already has saved.
    monthly_contribution (float): How much money the user can save each month.
    target_amount (float): The final financial goal.
    """

    # 1. Input Validation: Check for negative numbers
    if current_savings < 0 or monthly_contribution < 0 or target_amount < 0:
        return "❌ Error: Financial values cannot be negative. Please enter positive numbers."

    # 2. Input Validation: Check if they've already hit the goal
    if current_savings >= target_amount:
        return f"🎉 Congratulations! Your current savings (${current_savings:.2f}) already meet or exceed your target of ${target_amount:.2f}."

    # 3. Input Validation: Catch the "infinite loop" scenario
    # If they haven't met their goal and aren't contributing any money, they will never reach it
    if monthly_contribution <= 0:
        return "⚠️ Error: Monthly contribution must be greater than $0 to reach your savings goal."

    # 4. Core Logic: Simulate the passage of months
    months_needed = 0
    simulated_balance = current_savings

    # This loop keeps running as long as our simulated bank balance is less than the goal
    while simulated_balance < target_amount:
        simulated_balance += monthly_contribution
        months_needed += 1

    # 5. Build and return a user-friendly text report
    amount_left_to_save = target_amount - current_savings

    report = "=========================================\n"
    report += "          SAVINGS GOAL CALCULATOR        \n"
    report += "=========================================\n"
    report += f"🎯 Target Goal:    ${target_amount:.2f}\n"
    report += f"💰 Current Funds:  ${current_savings:.2f}\n"
    report += f"📉 Remaining Gap:  ${amount_left_to_save:.2f}\n"
    report += "-----------------------------------------\n"
    report += f"🕒 Timeline: It will take you {months_needed} months to reach your goal\n"
    report += f"   if you contribute ${monthly_contribution:.2f} every month.\n"
    report += "========================================="

    return report
````

## Gradio UI Integration

````python
import gradio as gr
import pandas as pd

# =====================================================================
# 1. CONNECTOR FUNCTIONS (Bridging your logic with Gradio)
# =====================================================================

def handle_file_analysis(file_obj):
    """
    Takes the uploaded file from Gradio, passes it through your
    cleaning, analysis, and recommendation pipeline.
    """
    if file_obj is None:
        return "⚠️ Please upload a CSV file first.", "⚠️ No data to analyze."

    # Run Function 1: Load and Clean (Gradio files provide a filepath in file_obj.name)
    cleaned_df = load_and_clean_transaction_data(file_obj.name)

    # If something went wrong during loading, exit early
    if cleaned_df is None or cleaned_df.empty:
        return "❌ Error: Could not process the file format.", "❌ Analysis aborted."

    # Run Function 2: Analyze spending patterns
    analysis_results = analyze_spending_patterns(cleaned_df)

    # Format a beautiful, readable text breakdown for the UI
    spending_report = f"💰 TOTAL OVERALL SPENDING: {analysis_results['Total Overall Spending']}\n"
    spending_report += f"🔴 TOP CATEGORY: {analysis_results['Top Category']}\n\n"
    spending_report += "📊 CATEGORY BREAKDOWN:\n"
    for cat, amt in analysis_results['Spending By Category'].items():
        spending_report += f" - {cat}: {amt}\n"

    # Run Function 3: Generate financial recommendations
    recommendations_report = generate_financial_recommendations(analysis_results)

    return spending_report, recommendations_report


def handle_savings_calculator(current, contribution, target):
    """
    Takes numeric inputs from Gradio text boxes/sliders and passes them
    into your savings calculator tool.
    """
    # Gradio inputs can pass as floats; call Function 4
    calculator_report = create_savings_calculator_tool(
        current_savings=float(current),
        monthly_contribution=float(contribution),
        target_amount=float(target)
    )
    return calculator_report

# =====================================================================
# 2. BUILDING THE GRADIO LAYOUT
# =====================================================================

def create_finance_assistant_ui():
    """
    Sets up the Gradio interface structure using simple tabs.
    """
    # Tab 1 Layout: Uploading and analyzing the CSV data
    dashboard_interface = gr.Interface(
        fn=handle_file_analysis,
        inputs=gr.File(label="Upload Student Transaction CSV", file_types=[".csv"]),
        outputs=[
            gr.Textbox(label="Spending Breakdown Report", lines=8),
            gr.Textbox(label="Smart Financial Recommendations", lines=12)
        ],
        description="Upload your Perth student transaction CSV file to get immediate, data-driven financial advice.",
        allow_flagging="never"
    )

    # Tab 2 Layout: Interactive savings timeline calculator
    calculator_interface = gr.Interface(
        fn=handle_savings_calculator,
        inputs=[
            gr.Number(label="Current Savings ($)", value=0),
            gr.Number(label="Monthly Contribution Target ($)", value=50),
            gr.Number(label="Target Savings Goal ($)", value=1000)
        ],
        outputs=gr.Textbox(label="Savings Goal Calculation Results", lines=10),
        description="Plan out your financial goals (e.g., car fund, emergency fund, textbook savings) to see your timeline.",
        allow_flagging="never"
    )

    # Combine both layouts into a single, professional tabbed application
    app = gr.TabbedInterface(
        interface_list=[dashboard_interface, calculator_interface],
        tab_names=["📊 Spending Dashboard", "🕒 Savings Goal Tracker"],
        title="🎓 SaverBuddy: Smart Student Finance Assistant"
    )

    return app
````

````python
# Create the interface object
finance_app = create_finance_assistant_ui()

# Launch the application!
# share=True creates a temporary public link you can open on your phone or show your class
finance_app.launch(debug=True)
````
